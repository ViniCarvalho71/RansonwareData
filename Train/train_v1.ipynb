## Imports

```python
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

## Importando a base de dados

```python
df = pd.read_csv('../Datasets/dataset_ransomware_v1.csv')[["num_sections","has_rsrc_section","max_entropy","avg_entropy","is_dll","is_64bit","is_ransomware"]]

df.head()
```

## Divisão dos dados

```python
features = df.iloc[:,0:6].values
features
```

```python
previsor = df.iloc[:,6:7].values
previsor
```

```python
x_train, x_test, y_train, Y_test = train_test_split(features,previsor, test_size=0.30, random_state=10)


alg_rf = RandomForestClassifier(n_estimators=500)
alg_rf.fit(x_train,y_train)
```

```python
Previsoes = alg_rf.predict(x_test)
matriz_de_confusao = confusion_matrix(y_test, Previsoes)
```

```python
plt.figure(figsize=(10,5))
sns.heatmap(matriz_de_confusao)
```
